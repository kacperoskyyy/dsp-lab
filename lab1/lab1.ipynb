```python
import numpy as np
import matplotlib.pyplot as plt

np.set_printoptions(linewidth=300)

x_mu = np.array([6, 2, 4, 3, 4, 5, 0, 0, 0, 0])
N = len(x_mu)

k = np.arange(N).reshape(N, 1)
mu = np.arange(N).reshape(1, N)

K = k @ mu

print("Matrix K:")
print(K)
print("-" * 50)

W = np.exp(1j * (2 * np.pi / N) * K)

print("Matrix W (rounded values for readability):")
print(np.round(W, 3))
print("-" * 50)

x_k = (1 / N) * (W @ x_mu)

print("Synthesized signal x_k (real and imaginary parts):")
print(np.round(x_k, 3))

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 8))

ax1.stem(np.arange(N), np.real(x_k), linefmt='b-', markerfmt='bo', basefmt='k-')
ax1.set_title(f'Synthesized signal x_k (N={N}) - Variant 11 (Real part)')
ax1.set_ylabel('Amplitude')
ax1.grid(True, linestyle='--', alpha=0.7)

ax2.stem(np.arange(N), np.imag(x_k), linefmt='r-', markerfmt='ro', basefmt='k-')
ax2.set_title(f'Synthesized signal x_k (N={N}) - Variant 11 (Imaginary part)')
ax2.set_xlabel('Sample index k')
ax2.set_ylabel('Amplitude')
ax2.grid(True, linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```
