```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Definition of vector x_mu for Variant 11 and its length N
x_mu = np.array([6, 2, 4, 3, 4, 5, 0, 0, 0, 0]).reshape(-1, 1)
N = len(x_mu)

# 2. Creating matrix K (containing products of k and mu)
k = np.arange(N).reshape(-1, 1)
mu = np.arange(N).reshape(1, -1)
K = k @ mu

# 3. Creating Fourier matrix W
W = np.exp(1j * (2 * np.pi / N) * K)

# 4. Synthesizing the signal using IDFT in matrix notation
x_k = (1 / N) * (W @ x_mu)

# Printing matrices K and W to the console
print("Matrix K:")
print(K)
print("\nMatrix W (values rounded for readability):")
print(np.round(W, 2))

# 5. Plotting the synthesized signal (Real and Imaginary parts)
plt.figure(figsize=(10, 5))

# Plot for the Real part
plt.subplot(1, 2, 1)
plt.stem(k.flatten(), np.real(x_k).flatten())
plt.title('Signal x[k] - Real Part')
plt.xlabel('Sample index (k)')
plt.ylabel('Amplitude')
plt.grid(True)

# Plot for the Imaginary part
plt.subplot(1, 2, 2)
plt.stem(k.flatten(), np.imag(x_k).flatten(), linefmt='orange', markerfmt='D')
plt.title('Signal x[k] - Imaginary Part')
plt.xlabel('Sample index (k)')
plt.grid(True)

plt.tight_layout()
plt.show()
```
